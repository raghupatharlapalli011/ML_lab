## Identifying Data Types and Preprocessing

```python
import numpy as np 
import pandas as pd 
from sklearn.datasets import load_diabetes 
from sklearn.preprocessing import LabelEncoder
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
col = ['Patient ID','Name','Age','Gender','Height','Weight','Blood Pressure','Glucose Level','Blood Group','Diabetes (Yes/No)']
df = pd.read_csv(url,names = col)
print(df.head())
print(df.dtypes)
numerical_df = df.select_dtypes(include = [np.number])
categorical_df = df.select_dtypes(exclude = [np.number])
print(numerical_df.head())
print(categorical_df.head())
label = LabelEncoder()
for i in col:
    df[i] = label.fit_transform(df[i])
df.drop(["Patient ID", "Name"], axis=1, inplace=True, errors="ignore")
print("\nCleaned Dataset")
print(df.head())
print("\nSummary Statistics")
print(df.describe())
```

## Handling Missing Values

```python
import numpy as np
import pandas as pd 
from sklearn.datasets import fetch_openml
titanic = fetch_openml('titanic', version=1, as_frame=True)
df = pd.DataFrame(data = titanic.data , columns = titanic.feature_names)
print(df.head())
col = titanic.feature_names
print(col)
missing_count = df.isnull().sum()
print("---------------")
print(missing_count)
print("filling null values with mean")
numerical_col = df.select_dtypes(include = [np.number]).columns
for i in  numerical_col:
    df[i] = df[i].fillna(df[i].mean())
categorical_df = df.select_dtypes(exclude = [np.number]).columns
print(categorical_df)
for i in categorical_df:
    df[i] = df[i].fillna(df[i].mode())
```
