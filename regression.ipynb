# Linear Regression

```python
# regression which is used to predict the numerical values have a single target variable 
import numpy as np 
import pandas as pd 
from sklearn.linear_model import LinearRegression
# predicting the marks based in the hours of the study
hours = [[1],[3],[5],[7],[9],[11]]
marks = [20 , 30 , 45 , 60 , 80 , 90]
model = LinearRegression()
model.fit(hours , marks)
predict = model.predict([[6]])
print(predict)
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

## Linear Regression based on the spliting the data 

```python
import numpy as np
import pandas as pd 
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
hours = [[1],[3],[5],[7],[9],[11]]
marks = [20 , 30 , 45 , 60 , 80 , 90]
X_train, X_test, y_train, y_test = train_test_split(
hours, marks, test_size=0.2, random_state=42)
model = LinearRegression()
model.fit(X_train,y_train)
predict = model.predict(X_test)
print(predict)
```

# Polynomial Regression

```python
import numpy as np 
import pandas as pd 
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
hours = [[1],[3],[5],[7],[9],[11]]
marks = [20 , 30 , 45 , 60 , 80 , 90]
poly_model = PolynomialFeatures(degree = 2)
x_poly = poly_model.fit_transform(hours)
model = LinearRegression()
model.fit(x_poly,marks)
predict = model.predict(poly_model.transform([[2]]))
print(predict)
```

## Polynomail Regression using the splitting the data

```python
import numpy as np 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures 
from sklearn.linear_model import LinearRegression
hours = [[1],[3],[5],[7],[9],[11]]
marks = [20 , 30 , 45 , 60 , 80 , 90]
X_train, X_test, y_train, y_test = train_test_split(
hours, marks, test_size=0.2, random_state=42)
poly_model = PolynomialFeatures(degree = 2)
x_poly = poly_model.fit_transform(X_train)
model = LinearRegression()
model.fit(x_poly , y_train)
predict = model.predict(poly_model.transform(X_test))
print(predict)
```

# logistic Regression

```python
# logistic regression for predicting the pass or fail depending on the marks
import numpy as np 
import pandas as pd 
from sklearn.linear_model import LogisticRegression
marks = [[20] , [30],[45],[60],[80],[90]]
result = [0 , 0 , 0 , 1 ,1 , 1 ]
model = LogisticRegression()
model.fit(marks , result)
predict = model.predict([[50]])
print(predict)
```

# Support vector machine

```python
import numpy as np 
import pandas as pd 
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
x = iris.data
y = iris.target
X_train, X_test, y_train, y_test = train_test_split(
x, y, test_size=0.2, random_state=42)
model = SVC(kernel='linear')
model.fit(X_train,y_train)
predict = model.predict(X_test)
accuracy = accuracy_score(y_test, predict)

print("Predicted Values:")
print(predict)

print("\nActual Values:")
print(y_test)
```
